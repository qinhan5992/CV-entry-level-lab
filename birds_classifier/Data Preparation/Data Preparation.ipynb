```python
import json

train_json_path = original_data_dir / "train_mini.json"
val_json_path = original_data_dir / "val.json"

with open(train_json_path, "r") as file:
    train_data = json.load(file)

with open(val_json_path, "r") as file:
    val_data = json.load(file)

print(train_data.keys())
print(val_data.keys())
```

```python
print("Training images:", len(train_data["images"]))
print("Training annotations:", len(train_data["annotations"]))
print("Training categories:", len(train_data["categories"]))

print()

print("Validation images:", len(val_data["images"]))
print("Validation annotations:", len(val_data["annotations"]))
print("Validation categories:", len(val_data["categories"]))
```

```python
category_information = {}

for category in train_data["categories"]:
    category_information[category["id"]] = category

print("Number of categories:", len(category_information))
```

```python
import random

random.seed(46)

all_category_ids = list(category_information.keys())
selected_category_ids = random.sample(all_category_ids, 500)

print("Number of selected species:", len(selected_category_ids))
print("First 10 selected category IDs:", selected_category_ids[:10])
```

```python
image_information = {}

for image in train_data["images"]:
    image_information[image["id"]] = image

print("Number of training images:", len(image_information))
```

```python
selected_train_images = []

for annotation in train_data["annotations"]:
    if annotation["category_id"] in selected_category_ids:
        image_id = annotation["image_id"]
        selected_train_images.append(image_information[image_id])

print("Number of selected training images:", len(selected_train_images))
print("First selected image:")
print(selected_train_images[0])
```

```python
val_image_information = {}

for image in val_data["images"]:
    val_image_information[image["id"]] = image

selected_val_images = []

for annotation in val_data["annotations"]:
    if annotation["category_id"] in selected_category_ids:
        image_id = annotation["image_id"]
        selected_val_images.append(val_image_information[image_id])

print("Number of selected validation images:", len(selected_val_images))
print("First selected validation image:")
print(selected_val_images[0])
```

```python
selected_test_images = selected_val_images

print("Number of selected test images:", len(selected_test_images))
```

```python
train_images_by_category = {}

for annotation in train_data["annotations"]:
    category_id = annotation["category_id"]

    if category_id in selected_category_ids:
        image_id = annotation["image_id"]
        image = image_information[image_id]

        if category_id not in train_images_by_category:
            train_images_by_category[category_id] = []

        train_images_by_category[category_id].append(image)

print("Number of selected categories:", len(train_images_by_category))
print("Images in the first category:", len(train_images_by_category[selected_category_ids[0]]))
```

```python
selected_training_images = []
selected_validation_images = []

random.seed(46)

for category_id in selected_category_ids:
    category_images = train_images_by_category[category_id]

    random.shuffle(category_images)

    selected_training_images.extend(category_images[:40])
    selected_validation_images.extend(category_images[40:50])

print("Number of final training images:", len(selected_training_images))
print("Number of final validation images:", len(selected_validation_images))
print("Number of final test images:", len(selected_test_images))
```

```python
train_image_category = {}
test_image_category = {}

for annotation in train_data["annotations"]:
    train_image_category[annotation["image_id"]] = annotation["category_id"]

for annotation in val_data["annotations"]:
    test_image_category[annotation["image_id"]] = annotation["category_id"]

print("Training image labels:", len(train_image_category))
print("Test image labels:", len(test_image_category))
```

```python
final_dataset_dir = Path(
    "/content/drive/MyDrive/COMP9517_Group_Assignment/data/selected_500_species_dataset"
)

final_train_dir = final_dataset_dir / "train"
final_val_dir = final_dataset_dir / "val"
final_test_dir = final_dataset_dir / "test"

final_train_dir.mkdir(parents=True, exist_ok=True)
final_val_dir.mkdir(parents=True, exist_ok=True)
final_test_dir.mkdir(parents=True, exist_ok=True)

print("Final dataset folders are ready.")
```

```python
training_images_file = original_data_dir / "training_images_for_mini_dataset.tar.gz"
validation_images_file = original_data_dir / "validation_images.tar.gz"

train_names = {image["file_name"] for image in selected_training_images}
val_names = {image["file_name"] for image in selected_validation_images}
test_names = {image["file_name"] for image in selected_test_images}

train_count = 0
val_count = 0
test_count = 0

with tarfile.open(training_images_file, "r:gz") as tar:
    for member in tar:
        original_name = member.name

        if original_name in train_names:
            member.name = Path(original_name).name
            tar.extract(member, final_train_dir)
            train_count += 1

        elif original_name in val_names:
            member.name = Path(original_name).name
            tar.extract(member, final_val_dir)
            val_count += 1

with tarfile.open(validation_images_file, "r:gz") as tar:
    for member in tar:
        original_name = member.name

        if original_name in test_names:
            member.name = Path(original_name).name
            tar.extract(member, final_test_dir)
            test_count += 1

print("Extraction completed.")
print("Training images extracted:", train_count)
print("Validation images extracted:", val_count)
print("Test images extracted:", test_count)
```

```python
def count_images(folder):
    image_count = 0

    for file in folder.iterdir():
        if file.suffix.lower() in [".jpg", ".jpeg", ".png"]:
            image_count += 1

    return image_count


print("Training images:", count_images(final_train_dir))
print("Validation images:", count_images(final_val_dir))
print("Test images:", count_images(final_test_dir))
```

```python
import csv

# Save image names and category labels.
def save_labels(images, label_information, output_file):
    with open(output_file, "w", newline="") as file:
        writer = csv.writer(file)
        writer.writerow(["file_name", "category_id"])

        for image in images:
            file_name = Path(image["file_name"]).name
            category_id = label_information[image["id"]]
            writer.writerow([file_name, category_id])


save_labels(
    selected_training_images,
    train_image_category,
    final_dataset_dir / "train_labels.csv"
)

save_labels(
    selected_validation_images,
    train_image_category,
    final_dataset_dir / "validation_labels.csv"
)

save_labels(
    selected_test_images,
    test_image_category,
    final_dataset_dir / "test_labels.csv"
)

print("Label files saved.")
```
