```python
from torchvision import transforms

IMAGE_SIZE = 224

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(
        IMAGE_SIZE,
        scale=(0.7, 1.0)
    ),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(15),
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2,
        hue=0.05
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

evaluation_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(IMAGE_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])
```

```python
from PIL import Image
from torch.utils.data import Dataset


class SpeciesDataset(Dataset):
    def __init__(self, dataframe, image_folder, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.image_folder = image_folder
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        file_name = row["file_name"]
        label = int(row["label"])

        image_path = os.path.join(
            self.image_folder,
            file_name
        )

        if not os.path.isfile(image_path):
            raise FileNotFoundError(
                f"Image not found: {image_path}"
            )

        try:
            image = Image.open(image_path).convert("RGB")
        except Exception as error:
            raise RuntimeError(
                f"Cannot read image: {image_path}"
            ) from error

        if self.transform is not None:
            image = self.transform(image)

        return image, label, file_name
```

```python
from torch.utils.data import DataLoader

debug_train_dataset = SpeciesDataset(
    dataframe=debug_train_df,
    image_folder=FULL_TRAIN_PATH,
    transform=train_transform
)

debug_val_dataset = SpeciesDataset(
    dataframe=debug_val_df,
    image_folder=FULL_VAL_PATH,
    transform=evaluation_transform
)

debug_test_dataset = SpeciesDataset(
    dataframe=debug_test_df,
    image_folder=FULL_TEST_PATH,
    transform=evaluation_transform
)

DEBUG_BATCH_SIZE = 8

debug_train_loader = DataLoader(
    debug_train_dataset,
    batch_size=DEBUG_BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True
)

debug_val_loader = DataLoader(
    debug_val_dataset,
    batch_size=DEBUG_BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

debug_test_loader = DataLoader(
    debug_test_dataset,
    batch_size=DEBUG_BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("Train dataset:", len(debug_train_dataset))
print("Validation dataset:", len(debug_val_dataset))
print("Test dataset:", len(debug_test_dataset))
```

##calculate_topk_correct

```python
def calculate_topk_correct(outputs, labels, topk=(1, 5)):
    """
    Count correct predictions for different top-k values.

    outputs:
        Model outputs with shape [batch_size, num_classes]

    labels:
        Correct labels with shape [batch_size]

    topk:
        Which top-k results to calculate
    """
    max_k = max(topk)

    # Get the class indices with the highest scores
    # Shape: [batch_size, max_k]
    _, predictions = outputs.topk(
        max_k,
        dim=1,
        largest=True,
        sorted=True
    )

    results = []

    for k in topk:
        # Keep only the first k predictions
        top_k_predictions = predictions[:, :k]

        # Check whether the correct label appears in top-k
        correct_k = (
            top_k_predictions
            .eq(labels.view(-1, 1))
            .any(dim=1)
            .sum()
            .item()
        )

        results.append(correct_k)

    return results
```

## train_one_epoch

```python
def train_one_epoch(
    model,
    data_loader,
    criterion,
    optimizer,
    device
):
    model.train()

    running_loss = 0.0
    top1_correct = 0
    top5_correct = 0
    total_samples = 0

    for images, labels, file_names in data_loader:
        images = images.to(
            device,
            non_blocking=True
        )

        labels = labels.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        batch_size = images.size(0)

        running_loss += loss.item() * batch_size

        batch_top1, batch_top5 = calculate_topk_correct(
            outputs=outputs,
            labels=labels,
            topk=(1, 5)
        )

        top1_correct += batch_top1
        top5_correct += batch_top5
        total_samples += batch_size

    epoch_loss = running_loss / total_samples
    top1_accuracy = top1_correct / total_samples
    top5_accuracy = top5_correct / total_samples

    return epoch_loss, top1_accuracy, top5_accuracy
```

##evaluate model


```python
def evaluate_model(
    model,
    data_loader,
    criterion,
    device
):
    model.eval()

    running_loss = 0.0
    top1_correct = 0
    top5_correct = 0
    total_samples = 0

    with torch.no_grad():
        for images, labels, file_names in data_loader:
            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )

            outputs = model(images)
            loss = criterion(outputs, labels)

            batch_size = images.size(0)

            running_loss += loss.item() * batch_size

            batch_top1, batch_top5 = calculate_topk_correct(
                outputs=outputs,
                labels=labels,
                topk=(1, 5)
            )

            top1_correct += batch_top1
            top5_correct += batch_top5
            total_samples += batch_size

    epoch_loss = running_loss / total_samples
    top1_accuracy = top1_correct / total_samples
    top5_accuracy = top5_correct / total_samples

    return epoch_loss, top1_accuracy, top5_accuracy
```

```python
history = {
    "train_loss": [],
    "train_accuracy": [],
    "val_loss": [],
    "val_accuracy": []
}

best_val_accuracy = 0.0
best_model_state = None

for epoch in range(DEBUG_EPOCHS):
    train_loss, train_accuracy = train_one_epoch(
        model=vit_model,
        data_loader=debug_train_loader,
        criterion=criterion,
        optimizer=optimizer,
        device=device
    )

    val_loss, val_accuracy = evaluate_model(
        model=vit_model,
        data_loader=debug_val_loader,
        criterion=criterion,
        device=device
    )

    history["train_loss"].append(train_loss)
    history["train_accuracy"].append(train_accuracy)
    history["val_loss"].append(val_loss)
    history["val_accuracy"].append(val_accuracy)

    if val_accuracy > best_val_accuracy:
        best_val_accuracy = val_accuracy

        best_model_state = {
            key: value.detach().cpu().clone()
            for key, value in vit_model.state_dict().items()
        }

    print(
        f"Epoch [{epoch + 1}/{DEBUG_EPOCHS}] "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Acc: {train_accuracy:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Acc: {val_accuracy:.4f}"
    )
```

```python
if best_model_state is not None:
    vit_model.load_state_dict(best_model_state)

print(f"Best validation accuracy: {best_val_accuracy:.4f}")
```

```python
# ============================================================
# Restore environment
# Restore paths, label files, mappings, datasets and loaders
# ============================================================

import os
import json
import random

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image


# ------------------------------------------------------------
# 1. Reproducibility
# ------------------------------------------------------------

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed(SEED)
    torch.cuda.manual_seed_all(SEED)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)


# ------------------------------------------------------------
# 2. Project paths
# ------------------------------------------------------------

PROJECT_ROOT = (
    "/content/gdrive/MyDrive/"
    "COMP9517_Group_Assignment_FULL"
)

DATA_ROOT = os.path.join(
    PROJECT_ROOT,
    "data"
)

FINAL_DATASET = os.path.join(
    DATA_ROOT,
    "selected_500_species_dataset"
)

FULL_TRAIN_PATH = os.path.join(
    FINAL_DATASET,
    "train"
)

FULL_VAL_PATH = os.path.join(
    FINAL_DATASET,
    "val"
)

FULL_TEST_PATH = os.path.join(
    FINAL_DATASET,
    "test"
)

TRAIN_CSV = os.path.join(
    FINAL_DATASET,
    "train_labels.csv"
)

VAL_CSV = os.path.join(
    FINAL_DATASET,
    "validation_labels.csv"
)

TEST_CSV = os.path.join(
    FINAL_DATASET,
    "test_labels.csv"
)

CHECKPOINT_DIR = os.path.join(
    PROJECT_ROOT,
    "checkpoints"
)

RESULTS_DIR = os.path.join(
    PROJECT_ROOT,
    "results"
)

os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)


# ------------------------------------------------------------
# 3. Check required paths
# ------------------------------------------------------------

required_paths = {
    "Project root": PROJECT_ROOT,
    "Data root": DATA_ROOT,
    "Dataset root": FINAL_DATASET,
    "Train images": FULL_TRAIN_PATH,
    "Validation images": FULL_VAL_PATH,
    "Test images": FULL_TEST_PATH,
    "Train CSV": TRAIN_CSV,
    "Validation CSV": VAL_CSV,
    "Test CSV": TEST_CSV
}

missing_paths = []

for path_name, path_value in required_paths.items():
    if not os.path.exists(path_value):
        missing_paths.append(
            f"{path_name}: {path_value}"
        )

if missing_paths:
    error_message = (
        "The following required paths were not found:\n"
        + "\n".join(missing_paths)
    )

    raise FileNotFoundError(error_message)


# ------------------------------------------------------------
# 4. Load CSV label files
# ------------------------------------------------------------

train_labels_df = pd.read_csv(TRAIN_CSV)
val_labels_df = pd.read_csv(VAL_CSV)
test_labels_df = pd.read_csv(TEST_CSV)

required_columns = {
    "file_name",
    "category_id"
}

for split_name, dataframe in [
    ("train", train_labels_df),
    ("validation", val_labels_df),
    ("test", test_labels_df)
]:
    missing_columns = (
        required_columns
        - set(dataframe.columns)
    )

    if missing_columns:
        raise ValueError(
            f"{split_name} CSV is missing columns: "
            f"{sorted(missing_columns)}"
        )


# ------------------------------------------------------------
# 5. Remove known invalid images
# ------------------------------------------------------------

INVALID_TRAIN_FILES = {
    "6de3c9f9-344a-470a-ac94-65ece211947b.jpg"
}

original_train_size = len(train_labels_df)

train_labels_df = train_labels_df[
    ~train_labels_df["file_name"].isin(
        INVALID_TRAIN_FILES
    )
].reset_index(drop=True)

removed_train_images = (
    original_train_size
    - len(train_labels_df)
)

print(
    "Removed known invalid train images:",
    removed_train_images
)


# ------------------------------------------------------------
# 6. Verify class sets
# ------------------------------------------------------------

train_categories = set(
    train_labels_df["category_id"].unique()
)

val_categories = set(
    val_labels_df["category_id"].unique()
)

test_categories = set(
    test_labels_df["category_id"].unique()
)

if train_categories != val_categories:
    raise ValueError(
        "Train and validation category sets are different."
    )

if train_categories != test_categories:
    raise ValueError(
        "Train and test category sets are different."
    )


# ------------------------------------------------------------
# 7. Create continuous labels: 0 to 499
# ------------------------------------------------------------

all_category_ids = sorted(train_categories)

category_to_label = {
    category_id: label
    for label, category_id
    in enumerate(all_category_ids)
}

label_to_category = {
    label: category_id
    for category_id, label
    in category_to_label.items()
}

NUM_CLASSES = len(category_to_label)

train_labels_df["label"] = (
    train_labels_df["category_id"]
    .map(category_to_label)
)

val_labels_df["label"] = (
    val_labels_df["category_id"]
    .map(category_to_label)
)

test_labels_df["label"] = (
    test_labels_df["category_id"]
    .map(category_to_label)
)

if (
    train_labels_df["label"].isna().any()
    or val_labels_df["label"].isna().any()
    or test_labels_df["label"].isna().any()
):
    raise ValueError(
        "Some category IDs could not be mapped."
    )


# ------------------------------------------------------------
# 8. Save category mapping
# ------------------------------------------------------------

MAPPING_PATH = os.path.join(
    FINAL_DATASET,
    "category_mapping.json"
)

mapping_data = {
    "category_to_label": {
        str(category_id): int(label)
        for category_id, label
        in category_to_label.items()
    },
    "label_to_category": {
        str(label): int(category_id)
        for label, category_id
        in label_to_category.items()
    }
}

with open(MAPPING_PATH, "w") as file:
    json.dump(
        mapping_data,
        file,
        indent=4
    )


# ------------------------------------------------------------
# 9. Image transformations
# ------------------------------------------------------------

IMAGE_SIZE = 224

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(
        IMAGE_SIZE,
        scale=(0.7, 1.0)
    ),
    transforms.RandomHorizontalFlip(
        p=0.5
    ),
    transforms.RandomRotation(
        degrees=15
    ),
    transforms.ColorJitter(
        brightness=0.2,
        contrast=0.2,
        saturation=0.2,
        hue=0.05
    ),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

evaluation_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(IMAGE_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ------------------------------------------------------------
# 10. Custom Dataset class
# ------------------------------------------------------------

class SpeciesDataset(Dataset):
    def __init__(
        self,
        dataframe,
        image_folder,
        transform=None
    ):
        self.dataframe = (
            dataframe
            .reset_index(drop=True)
            .copy()
        )

        self.image_folder = image_folder
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        file_name = row["file_name"]
        label = int(row["label"])

        image_path = os.path.join(
            self.image_folder,
            file_name
        )

        if not os.path.isfile(image_path):
            raise FileNotFoundError(
                f"Image not found: {image_path}"
            )

        try:
            with Image.open(image_path) as image:
                image = image.convert("RGB")
        except Exception as error:
            raise RuntimeError(
                f"Cannot read image: {image_path}"
            ) from error

        if self.transform is not None:
            image = self.transform(image)

        return image, label, file_name


# ------------------------------------------------------------
# 11. Create datasets
# ------------------------------------------------------------

train_dataset = SpeciesDataset(
    dataframe=train_labels_df,
    image_folder=FULL_TRAIN_PATH,
    transform=train_transform
)

val_dataset = SpeciesDataset(
    dataframe=val_labels_df,
    image_folder=FULL_VAL_PATH,
    transform=evaluation_transform
)

test_dataset = SpeciesDataset(
    dataframe=test_labels_df,
    image_folder=FULL_TEST_PATH,
    transform=evaluation_transform
)


# ------------------------------------------------------------
# 12. Create DataLoaders
# ------------------------------------------------------------

BATCH_SIZE = 16

# Use 0 for more stable reading directly from Google Drive
NUM_WORKERS = 0

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available()
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=NUM_WORKERS,
    pin_memory=torch.cuda.is_available()
)


# ------------------------------------------------------------
# 13. Final summary
# ------------------------------------------------------------

print("=" * 60)
print("Environment restored successfully")
print("=" * 60)

print("Device:", device)

if torch.cuda.is_available():
    print(
        "GPU:",
        torch.cuda.get_device_name(0)
    )

print("\nProject paths:")
print("Project root:", PROJECT_ROOT)
print("Data root:", DATA_ROOT)
print("Dataset root:", FINAL_DATASET)

print("\nImage paths:")
print("Train:", FULL_TRAIN_PATH)
print("Validation:", FULL_VAL_PATH)
print("Test:", FULL_TEST_PATH)

print("\nDataset sizes:")
print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))

print("\nNumber of classes:", NUM_CLASSES)

print(
    "Label range:",
    int(train_labels_df["label"].min()),
    "to",
    int(train_labels_df["label"].max())
)

print("\nDataLoader batches:")
print("Train:", len(train_loader))
print("Validation:", len(val_loader))
print("Test:", len(test_loader))

print("\nDataLoader settings:")
print("Batch size:", BATCH_SIZE)
print("Number of workers:", NUM_WORKERS)

print("\nCheckpoint directory:")
print(CHECKPOINT_DIR)

print("\nResults directory:")
print(RESULTS_DIR)

print("\nMapping saved to:")
print(MAPPING_PATH)
```

```python
import time

history = {
    "train_loss": [],
    "train_top1": [],
    "train_top5": [],
    "val_loss": [],
    "val_top1": [],
    "val_top5": [],
    "epoch_time": []
}

best_val_top1 = -1.0
best_model_state = None

for epoch in range(HEAD_EPOCHS):
    start_time = time.time()

    train_loss, train_top1, train_top5 = train_one_epoch(
        model=pretrained_vit,
        data_loader=train_loader,
        criterion=criterion,
        optimizer=optimizer,
        device=device
    )

    val_loss, val_top1, val_top5 = evaluate_model(
        model=pretrained_vit,
        data_loader=val_loader,
        criterion=criterion,
        device=device
    )

    epoch_time = time.time() - start_time

    history["train_loss"].append(train_loss)
    history["train_top1"].append(train_top1)
    history["train_top5"].append(train_top5)
    history["val_loss"].append(val_loss)
    history["val_top1"].append(val_top1)
    history["val_top5"].append(val_top5)
    history["epoch_time"].append(epoch_time)

    if val_top1 > best_val_top1:
        best_val_top1 = val_top1

        best_model_state = {
            key: value.detach().cpu().clone()
            for key, value in pretrained_vit.state_dict().items()
        }

    print(
        f"Epoch [{epoch + 1}/{HEAD_EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Top-1: {train_top1:.4f} | "
        f"Train Top-5: {train_top5:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Top-1: {val_top1:.4f} | "
        f"Val Top-5: {val_top5:.4f} | "
        f"Time: {epoch_time / 60:.2f} min"
    )
```

##find the dameged image

```python
from PIL import Image
from tqdm.auto import tqdm
import os

def find_invalid_images(dataframe, image_folder):
    invalid_files = []

    for file_name in tqdm(
        dataframe["file_name"],
        desc=f"Checking {os.path.basename(image_folder)}"
    ):
        image_path = os.path.join(image_folder, file_name)

        try:
            # 先检查文件是否存在且大小大于0
            if not os.path.isfile(image_path):
                invalid_files.append(file_name)
                continue

            if os.path.getsize(image_path) == 0:
                invalid_files.append(file_name)
                continue

            # 再检查是否是有效图片
            with Image.open(image_path) as image:
                image.verify()

        except Exception:
            invalid_files.append(file_name)

    return invalid_files


invalid_train_files = find_invalid_images(
    train_labels_df,
    FULL_TRAIN_PATH
)

invalid_val_files = find_invalid_images(
    val_labels_df,
    FULL_VAL_PATH
)

invalid_test_files = find_invalid_images(
    test_labels_df,
    FULL_TEST_PATH
)

print("Invalid train images:", len(invalid_train_files))
print("Invalid validation images:", len(invalid_val_files))
print("Invalid test images:", len(invalid_test_files))

print("\nTrain invalid examples:")
print(invalid_train_files[:10])
```

##Unfroze

```python
# Freeze all parameters
for parameter in pretrained_vit.parameters():
    parameter.requires_grad = False

# Unfroze last three Transformer Encoder blocks
for block in pretrained_vit.encoder.layers[-3:]:
    for parameter in block.parameters():
        parameter.requires_grad = True

# Unfroze LayerNorm
for parameter in pretrained_vit.encoder.ln.parameters():
    parameter.requires_grad = True

# Unfroze classification head
for parameter in pretrained_vit.heads.parameters():
    parameter.requires_grad = True
```

```python
import time

LATEST_FINE_TUNE_PATH = os.path.join(
    CHECKPOINT_DIR,
    "vit_b16_partial_finetune_latest.pth"
)

BEST_FINE_TUNE_PATH = os.path.join(
    CHECKPOINT_DIR,
    "vit_b16_partial_finetune_best.pth"
)

fine_tune_history = {
    "train_loss": [],
    "train_top1": [],
    "train_top5": [],
    "val_loss": [],
    "val_top1": [],
    "val_top5": [],
    "epoch_time": []
}

best_val_top1 = -1.0
start_epoch = 0

for epoch in range(start_epoch, FINE_TUNE_EPOCHS):
    epoch_start_time = time.time()

    train_loss, train_top1, train_top5 = train_one_epoch(
        model=pretrained_vit,
        data_loader=train_loader,
        criterion=criterion,
        optimizer=optimizer,
        device=device
    )

    val_loss, val_top1, val_top5 = evaluate_model(
        model=pretrained_vit,
        data_loader=val_loader,
        criterion=criterion,
        device=device
    )

    epoch_time = time.time() - epoch_start_time

    fine_tune_history["train_loss"].append(train_loss)
    fine_tune_history["train_top1"].append(train_top1)
    fine_tune_history["train_top5"].append(train_top5)
    fine_tune_history["val_loss"].append(val_loss)
    fine_tune_history["val_top1"].append(val_top1)
    fine_tune_history["val_top5"].append(val_top5)
    fine_tune_history["epoch_time"].append(epoch_time)

    is_best = val_top1 > best_val_top1

    if is_best:
        best_val_top1 = val_top1

    checkpoint_data = {
        "epoch": epoch + 1,
        "model_state_dict": pretrained_vit.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "best_val_top1": best_val_top1,
        "history": fine_tune_history,
        "category_to_label": category_to_label,
        "label_to_category": label_to_category,
        "num_classes": NUM_CLASSES,
        "training_stage": "last_3_encoder_blocks_finetuned"
    }

    torch.save(
        checkpoint_data,
        LATEST_FINE_TUNE_PATH
    )

    if is_best:
        torch.save(
            checkpoint_data,
            BEST_FINE_TUNE_PATH
        )

    print(
        f"Epoch [{epoch + 1}/{FINE_TUNE_EPOCHS}] | "
        f"Train Loss: {train_loss:.4f} | "
        f"Train Top-1: {train_top1:.4f} | "
        f"Train Top-5: {train_top5:.4f} | "
        f"Val Loss: {val_loss:.4f} | "
        f"Val Top-1: {val_top1:.4f} | "
        f"Val Top-5: {val_top5:.4f} | "
        f"Time: {epoch_time / 60:.2f} min"
    )

    print("Latest checkpoint saved.")

    if is_best:
        print("New best checkpoint saved.")
```

```python
directed_confusions = []

for true_label in range(NUM_CLASSES):
    for predicted_label in range(NUM_CLASSES):
        if true_label == predicted_label:
            continue

        confusion_count = int(
            cm[true_label, predicted_label]
        )

        if confusion_count == 0:
            continue

        true_total = int(
            cm[true_label].sum()
        )

        confusion_rate = (
            confusion_count / true_total
            if true_total > 0
            else 0.0
        )

        directed_confusions.append({
            "true_label": true_label,
            "true_category_id": int(
                label_to_category[true_label]
            ),
            "predicted_label": predicted_label,
            "predicted_category_id": int(
                label_to_category[predicted_label]
            ),
            "confusion_count": confusion_count,
            "true_class_total": true_total,
            "confusion_rate": confusion_rate
        })

directed_confusions_df = pd.DataFrame(
    directed_confusions
).sort_values(
    by=[
        "confusion_count",
        "confusion_rate"
    ],
    ascending=False
).reset_index(drop=True)

display(
    directed_confusions_df.head(20)
)
```

```python
pair_confusions = []

for label_a in range(NUM_CLASSES):
    for label_b in range(
        label_a + 1,
        NUM_CLASSES
    ):
        a_to_b = int(
            cm[label_a, label_b]
        )

        b_to_a = int(
            cm[label_b, label_a]
        )

        total_confusion = (
            a_to_b + b_to_a
        )

        if total_confusion == 0:
            continue

        pair_confusions.append({
            "label_a": label_a,
            "category_id_a": int(
                label_to_category[label_a]
            ),
            "label_b": label_b,
            "category_id_b": int(
                label_to_category[label_b]
            ),
            "a_predicted_as_b": a_to_b,
            "b_predicted_as_a": b_to_a,
            "total_bidirectional_confusion": (
                total_confusion
            )
        })

pair_confusions_df = pd.DataFrame(
    pair_confusions
).sort_values(
    by="total_bidirectional_confusion",
    ascending=False
).reset_index(drop=True)

display(
    pair_confusions_df.head(20)
)
```

```python
import time
import torch

pretrained_vit.eval()

total_images = 0

# GPU warm-up
with torch.no_grad():
    for images, labels, file_names in test_loader:
        images = images.to(device)

        for _ in range(3):
            _ = pretrained_vit(images)

        break

if torch.cuda.is_available():
    torch.cuda.synchronize()

start_time = time.perf_counter()

with torch.no_grad():
    for images, labels, file_names in test_loader:
        images = images.to(
            device,
            non_blocking=True
        )

        _ = pretrained_vit(images)

        total_images += images.size(0)

if torch.cuda.is_available():
    torch.cuda.synchronize()

end_time = time.perf_counter()

total_inference_time = end_time - start_time
time_per_image = total_inference_time / total_images
throughput = total_images / total_inference_time

print(f"Total test images: {total_images}")
print(
    f"Total inference time: "
    f"{total_inference_time:.2f} seconds"
)
print(
    f"Inference time per image: "
    f"{time_per_image * 1000:.2f} ms"
)
print(
    f"Throughput: "
    f"{throughput:.2f} images/second"
)
```
